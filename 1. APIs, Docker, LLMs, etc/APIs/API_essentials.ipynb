### HTTP methods-

All website use HTTP to communicate between the server and client software to display and access the websites/webpages. 

You communicate with server that you want to perform CRUD (Create Retrieve Update Delete) action-

| Method | Description |
|--------|-------------|
| `GET` | This method will RETRIEVE resources |
| `POST` | This method will CREATE resources |
| `PUT` | This method will UPDATE resources |
| `DELETE` | This method will DELETE resources |

### Path params

path parameters are dynamic segments of URL path used to identify a specific resources.

### HTTP codes-

- 2XX       = Success
    - 200       OK
    - 201       Created
    - 204       No content (Sucess but no data returned)       
- 3XX       = Redirection

- 4XX       = Client Error
    - 400       Malformed or invalid request
    - 401       Unauthorized request
    - 403       Forbidden
    - 404       Resource doesn't exists
- 5XX       = Server Error
    - 500       Internal Server error
    - 502       Bad gateway
    - 503       Service is down or unavailable

### HTTPException 

is a special exception in FASTAPI used to return custom HTTP error responsses when something goes wrong in your API. Instead of return normal JSON or cashing the server, you can gracefully raise an error with:
- a proper HTTP status code (like 404, 400, 403 etc)
- a custom error message
- (optional) extra headers

### Query params

are optional keys-value pairs appended to the end of hte URL used to pass additional data to the server in HTTP request. They are typically employed for operations like filtering, sorting searching and pagination. without altering the endpoint path itself.

`/view?city=Delhi%sort_by=age`

| Symbol | Meaning |
|--------|---------|
| `?` | Start of query |
| `key=value` | Each parameter is a key-value pair |
| `&` | Separates multiple parameters |
