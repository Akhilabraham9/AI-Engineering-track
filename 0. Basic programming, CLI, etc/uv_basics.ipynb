# UV commands -

- `uv init`                                                     - initialized files
- `uv venve <virtual environment name>`                         - Creates a virtual environment
- `.<virtual environment name>\Scripts\activate`                - activates the virtual environment 
- `uv add <1 or more (no comma) libraries name>`                - installs libraries (same as `pip install <libraries>`)
- `deactivate`                                                  - deactivates the environment
- `$env:UV_PROJECT_ENVIRONMENT = '<virtual environment name'>`  - changes the virtual environment to what we specify
