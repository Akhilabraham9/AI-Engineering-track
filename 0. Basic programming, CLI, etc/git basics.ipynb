# GIT commands

1. Git initialize                           - `git init`
2. Git add files/folders                    - `git add .`                                   or `git add <filename>`
3. Git commit changes                       - `git commit -m "<commit message>`
4. Git add the link to github repositor     - `git add origin <link copied from github>`
5. Git rename exisiting branch name         - `git branch -M <name>`
6. Git switch branches/ create new branch   - `git checkout <other branch>`                 or `git switch -c <other branch>`

7. Git overwrite exisiting branch
    - `git checkout <branch which is going to be overwritten>`  - change to branch which should be overwritten
    - `git reset --hard <branch which will overwrite>`          - overwrite the `branch 1` with `branch 2`
    - `git push origin main --force`                            - push the branch to commit the changes
